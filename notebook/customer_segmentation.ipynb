```python
def remove_iqr_outliers(data, column):
    Q1 = data[column].quantile(0.25)
    Q3 = data[column].quantile(0.75)
    IQR = Q3 - Q1
    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR
    return data[
        (data[column] >= lower_bound) &
        (data[column] <= upper_bound)
    ]
```

```python
customer_model_df = customer_df.copy()
for column in ['Monetary', 'Frequency', 'Recency']:
    customer_model_df = remove_iqr_outliers(
        customer_model_df,
        column
    )
```

```python
print("Original customers:", len(customer_df))
print("Customers after outlier treatment:", len(customer_model_df))
```

```python
features = [
    'Recency',
    'Frequency',
    'Monetary'
]
```

```python
from sklearn.preprocessing import StandardScaler
scaler = StandardScaler()
scaled_features = scaler.fit_transform(
    customer_model_df[features]
)
```

```python
from sklearn.cluster import KMeans
```

```python
inertia = []
for k in range(2, 11):
    kmeans = KMeans(
        n_clusters=k,
        random_state=42,
        n_init=10
    )
    kmeans.fit(scaled_features)
    inertia.append(kmeans.inertia_)
```

```python
plt.figure(figsize=(8, 5))
plt.plot(
    range(2, 11),
    inertia,
    marker='o'
)
plt.xlabel('Number of Clusters')
plt.ylabel('Within-Cluster Sum of Squares')
plt.title('Elbow Method for Selecting Number of Clusters')
plt.tight_layout()
plt.savefig('../visuals/elbow_curve.png')
plt.show()
```

```python
from sklearn.metrics import silhouette_score
```

```python
silhouette_scores = {}
for k in range(2, 11):
    kmeans = KMeans(
        n_clusters=k,
        random_state=42,
        n_init=10
    )
    labels = kmeans.fit_predict(scaled_features)
    score = silhouette_score(
        scaled_features,
        labels
    )
    silhouette_scores[k] = score
    print(
        f"Clusters: {k}, "
        f"Silhouette Score: {score:.4f}"
    )
```

```python
plt.figure(figsize=(8, 5))
plt.plot(
    list(silhouette_scores.keys()),
    list(silhouette_scores.values()),
    marker='o'
)
plt.xlabel('Number of Clusters')
plt.ylabel('Silhouette Score')
plt.title('Silhouette Score by Number of Clusters')
plt.tight_layout()
plt.savefig('../visuals/silhouette_scores.png')
plt.show()
```

```python
optimal_k = 3
kmeans = KMeans(
    n_clusters=optimal_k,
    random_state=42,
    n_init=10
)
customer_model_df['Cluster'] = kmeans.fit_predict(
    scaled_features
)
```

```python
cluster_summary = (
    customer_model_df
    .groupby('Cluster')
    [['Recency', 'Frequency', 'Monetary', 'TotalQuantity']]
    .mean()
)
cluster_summary
```

```python
cluster_names = {
    0: 'At-Risk Customers',
    1: 'Premium Customers',
    2: 'Regular Customers'
}
customer_model_df['Segment'] = (
    customer_model_df['Cluster']
    .map(cluster_names)
)
```
